## Conversor de texto crudo en métricas
- Convención snake_case: Variables en minúsculas con guiones bajos
- Casteo explícito: `float()` habilita operaciones matemáticas
- Limpieza previa: `.split()` elimina espacios ocultos

```python
# Dato exportado como texto
inversion_raw = "1250.50"
clics = 4250

# Casteo a float para operar
inversion_eur = float(inversion_raw)
cpc = round(inversion_eur / clics, 2)
print(f"CPC: {cpc}€")
```

### Ejercicio "Entrada de datos"
Crea un programa que pregunte al usuario:
- Su nombre
- El número de visitas que ha recibido una página web
- El número de compras realizadas

El programa debe mostrar los datos recibidos y la tasa de conversión

```python
name = input("Dime nombre de la página: ")
visitas = float(input("Dime el número de visitas: "))
compras = float(input("Dime ventas: "))

tasa = round(visitas/compras, 2)
print(f"Nombre: {name}\nNúmero de visitas: {visitas}\nCompras realizadas: {compras}")
print(f"Tasa de Conversión: {tasa}")
```

### EJEMPLO: Calculo de Metricas de Marketing Digital
Realizado por Daniela(Tutora)

```python
campana = "Black_Friday_2026"       # str: Identificador
impresiones = 150000                # int: Veces que se mostró el anuncio
clics = 4250                        # int: Clics obtenidos
inversion_raw = "1250.50"           # str: Dato crudo exportado de una plataforma

# 1. Transformación de tipo (Casting) para poder operar
inversion_eur = float(inversion_raw)

# 2. Cálculo de KPIs de negocio
ctr = (clics / impresiones) * 100                  # Click-Through Rate (%)
cpc = round(inversion_eur / clics, 2)              # Coste por Clic (€)
es_rentable = ctr > 2.5 and cpc < 0.35             # bool: Regla de negocio

print(f"Campaña: {campana} | CTR: {ctr:.2f}% | CPC: {cpc}€ | ¿Objetivo cumplido?: {es_rentable}")
```
