3. Calcula la `media_ventas` diaria dividiendo `total_ventas` entre el número de días (`len(ventas_semana)`), redondeada a 2 decimales.

**Contexto:** En un experimento A/B de una página web, uno de los usuarios cerró el navegador antes de que se registrara su tiempo de permanencia (`None`). Si intentas promediar la columna directamente, Python dará error.

**Tu tarea:**
1. Crea una lista vacía llamada `tiempos_validos`.
2. Recorre la tabla `experimento_ab` y guarda en `tiempos_validos` únicamente los valores de `"tiempo_seg"` que **no sean `None`** (`if fila["tiempo_seg"] is not None:`).
3. Calcula el tiempo medio de permanencia con esos registros válidos (redondeado a 2 decimales).

> ✅ **Resultado esperado:** `Tiempos válidos: [40, 95, 110] | Tiempo medio: 81.67 seg`

````python
# Resultados crudos de Test A/B
experimento_ab = [
    {"user": "U1", "grupo": "A", "tiempo_seg": 40, "convirtio": False},
    {"user": "U2", "grupo": "B", "tiempo_seg": 95, "convirtio": True},
    {"user": "U3", "grupo": "A", "tiempo_seg": None, "convirtio": False},
    {"user": "U4", "grupo": "B", "tiempo_seg": 110, "convirtio": True},
]

# 1. Creamos la lista vacía para almacenar los tiempos válidos
tiempos_validos = []

# 2. Recorremos la tabla del experimento para filtrar nulos
for fila in experimento_ab:
    if fila["tiempo_seg"] is not None:
        tiempos_validos.append(fila["tiempo_seg"])

# 3. Calculamos la media de tiempo de los datos filtrados
tiempo_medio = round(sum(tiempos_validos) / len(tiempos_validos), 2)

# Imprimimos los resultados en el formato esperado
print(f"Tiempos válidos: {tiempos_validos} | Tiempo medio: {tiempo_medio} seg")
````

### 🔴 Ejercicio 2.5 — Agrupación y Resumen por Grupo con Diccionarios *(Reto Integrador)*
**Contexto:** En analítica de datos necesitamos agrupar filas por una categoría (como haría un `GROUP BY` en SQL) para comparar métricas entre grupos. Vas a resolver el análisis completo del Test A/B usando **únicamente diccionarios y listas nativas de Python, sin importar ninguna librería**.

**Tu tarea:**
1. Crea un diccionario acumulador llamado `resumen_grupos` con la estructura inicial:
   ```python
   resumen_grupos = {
       "A": {"usuarios": 0, "convertidos": 0, "tiempos": []},
       "B": {"usuarios": 0, "convertidos": 0, "tiempos": []},
   }
   ```
2. Recorre la lista `experimento_ab` con un bucle `for`:
   * Obtén el grupo de la fila (`grupo = fila["grupo"]`).
   * Suma `1` al contador `resumen_grupos[grupo]["usuarios"]`.
   * Si `fila["convirtio"]` es `True`, suma `1` a `resumen_grupos[grupo]["convertidos"]`.
   * Si `fila["tiempo_seg"]` **no es `None`**, añade ese tiempo a la lista `resumen_grupos[grupo]["tiempos"]`.
3. Crea un diccionario final `resultado_ab = {}` recorriendo `resumen_grupos.items()` para calcular en cada grupo:
   * `tasa_conv_pct`: `(convertidos / usuarios) * 100`, redondeada a 1 decimal.
   * `tiempo_medio_seg`: promedio de la lista `tiempos`, redondeado a 1 decimal.
4. Imprime el resumen comparativo del Grupo A frente al Grupo B.

> 💡 **Pista:** Accede a los valores internos encadenando corchetes, por ejemplo: `resumen_grupos[grupo]["usuarios"] += 1`.
> ✅ **Resultado esperado:** `Grupo A: 0.0% conversión (40.0s) | Grupo B: 100.0% conversión (102.5s)`

````python
# Usamos la misma tabla experimento_ab (sin importar librerías externas)
experimento_ab = [
    {"user": "U1", "grupo": "A", "tiempo_seg": 40, "convirtio": False},
    {"user": "U2", "grupo": "B", "tiempo_seg": 95, "convirtio": True},
    {"user": "U3", "grupo": "A", "tiempo_seg": None, "convirtio": False},
    {"user": "U4", "grupo": "B", "tiempo_seg": 110, "convirtio": True},
]

# 1. Creamos el diccionario acumulador estructurado por grupo
resumen_grupos = {
    "A": {"usuarios": 0, "convertidos": 0, "tiempos": []},
    "B": {"usuarios": 0, "convertidos": 0, "tiempos": []},
}

# 2. Recorremos la tabla experimento_ab para agrupar los datos
for fila in experimento_ab:
    grupo = fila["grupo"]

    # Sumamos un usuario al grupo correspondiente
    resumen_grupos[grupo]["usuarios"] += 1

    # Si convirtió, sumamos uno al contador de conversiones de ese grupo
    if fila["convirtio"]:
        resumen_grupos[grupo]["convertidos"] += 1

    # Si tiene un tiempo válido, lo añadimos a su lista de tiempos
    if fila["tiempo_seg"] is not None:
        resumen_grupos[grupo]["tiempos"].append(fila["tiempo_seg"])

# 3. Calculamos las métricas finales (tasas y promedios) para cada grupo
resultado_ab = {}
for grupo, datos in resumen_grupos.items():
    usuarios = datos["usuarios"]
    convertidos = datos["convertidos"]
    tiempos = datos["tiempos"]

    # Calculamos porcentaje de conversión
    tasa_conv_pct = round((convertidos / usuarios) * 100, 1) if usuarios > 0 else 0.0

    # Calculamos la media de tiempo de permanencia
    tiempo_medio_seg = round(sum(tiempos) / len(tiempos), 1) if len(tiempos) > 0 else 0.0

    # Guardamos el reporte final procesado para el grupo
    resultado_ab[grupo] = {
        "tasa_conversion": f"{tasa_conv_pct}%",
        "tiempo_medio": f"{tiempo_medio_seg}s"
    }

# 4. Imprimimos el resumen comparativo final en pantalla
print(f"Grupo A: {resultado_ab['A']['tasa_conversion']} conversión ({resultado_ab['A']['tiempo_medio']}) | "
      f"Grupo B: {resultado_ab['B']['tasa_conversion']} conversión ({resultado_ab['B']['tiempo_medio']})")
````

### 🔴 Ejercicio 3.5 — Agregación por Categoría + Anonimización GDPR *(Reto Integrador)*
**Contexto:** Vas a compartir un resumen de ventas con un equipo externo, pero por normativa de protección de datos (**GDPR**) está prohibido exportar correos electrónicos reales sin anonimizar. Además, las categorías están escritas de forma inconsistente (`" electrónica "` vs `"ELECTRÓNICA"`).

**Tu tarea:**
1. Crea un diccionario vacío `ingresos_por_categoria = {}` y una lista vacía `emails_anonimizados = []`.
2. Recorre `transacciones`:
   * Normaliza la categoría con `.strip().title()`.
   * Acumula el importe en `ingresos_por_categoria` (pista: `ingresos_por_categoria[cat] = ingresos_por_categoria.get(cat, 0.0) + t["importe"]`).
   * Anonimiza el email dejando visible **solo la primera letra** del usuario, seguida de `"***@"` y el dominio (ej. `"ana.perez@mail.com"` $
ightarrow$ `"a***@mail.com"`). Guárdalo en `emails_anonimizados`.
3. Imprime el diccionario de ingresos por categoría y la lista de correos protegidos.

> 💡 **Pista:** `usuario, dominio = email.split("@")` separa el correo en dos partes usando la arroba como corte.
> ✅ **Resultado esperado:** `{'Electrónica': 350.0, 'Hogar': 100.0}` y `['a***@mail.com', 'l***@mail.com', 's***@mail.com', 'm***@mail.com']`

````python
# Transacciones con categorías sucias y datos personales (PII)
transacciones = [
    {"categoria": " electrónica ", "importe": 150.0, "email": "ana.perez@mail.com"},
    {"categoria": "Hogar", "importe": 60.0, "email": "luis.gomez@mail.com"},
    {"categoria": "ELECTRÓNICA", "importe": 200.0, "email": "sara.diaz@mail.com"},
    {"categoria": "  hogar", "importe": 40.0, "email": "marta.ruiz@mail.com"},
]

# 1. Inicializamos los contenedores requeridos.
# Usamos un diccionario para acumular ingresos por categoría y una lista para los correos protegidos.
ingresos_por_categoria = {}
emails_anonimizados = []

# 2. Iteramos por cada transacción en el dataset
for t in transacciones:
    # --- PASO A: Normalización de Categorías ---
    # Eliminamos espacios en blanco con .strip() y capitalizamos correctamente con .title()
    # para evitar duplicidades debido a diferencias de formato (ej. "  hogar" vs "Hogar").
    cat_limpia = t["categoria"].strip().title()

    # --- PASO B: Agregación de Ingresos ---
    # Acumulamos el importe de forma dinámica.
    # Usamos .get() con valor de respaldo 0.0 para inicializar la categoría la primera vez que aparezca.
    ingresos_por_categoria[cat_limpia] = ingresos_por_categoria.get(cat_limpia, 0.0) + t["importe"]

    # --- PASO C: Anonimización conforme a GDPR ---
    # Separamos el correo en dos partes usando la '@' como delimitador:
    # 'usuario' (ej. 'ana.perez') y 'dominio' (ej. 'mail.com').
    usuario, dominio = t["email"].split("@")

    # Extraemos únicamente el primer carácter del usuario ('usuario[0]'),
    # agregamos una máscara de privacidad '***' y lo volvemos a unir con el dominio.
    email_protegido = f"{usuario[0]}***@{dominio}"

    # Añadimos el correo anonimizado a nuestra lista final para asegurar la confidencialidad.
    emails_anonimizados.append(email_protegido)

# 3. Imprimimos el reporte final para validar que las métricas y la privacidad sean correctas
print("Ingresos por categoría limpios:", ingresos_por_categoria)
print("Emails anonimizados (GDPR Compliance):", emails_anonimizados)
````

2. Llama a tu función con `100` (usará el IVA general del 21 % por defecto) y guarda el resultado en `precio_general`.

### 🟢 Ejercicio 4.2 — Función Defensiva de Limpieza de Texto *(Básico)*
**Contexto:** Cuando limpias columnas de texto en datasets reales, tarde o temprano aparece un valor nulo (`None`). Si llamas a `.strip()` sobre `None`, el programa se rompe (`AttributeError`).

**Tu tarea:**
1. Crea una función `normalizar_texto(texto)` que:
   * Si `texto is None` (o está vacío `""`), devuelva `"Sin Dato"`.
   * En caso contrario, convierta `texto` a `str`, elimine espacios laterales y devuelva el texto en formato Título (`.strip().title()`).
2. Prueba tu función con `"  barcelona "`, `"MADRID"` y `None`.

> ✅ **Resultado esperado:** `'Barcelona'`, `'Madrid'`, `'Sin Dato'`

````python
def normalizar_texto(texto):
    # --- POR QUÉ ---
    # Verificamos si es None o vacío primero. Si intentamos aplicar métodos como .strip()
    # sobre un tipo None, Python lanzará un error fatal (AttributeError) deteniendo el flujo.
    if texto is None or str(texto).strip() == "":
        return "Sin Dato"

    # --- POR QUÉ ---
    # Si el dato es válido, lo forzamos a cadena por seguridad (casteo), eliminamos
    # espacios laterales (.strip()) y lo formateamos como título (.title()) para unificar categorías.
    return str(texto).strip().title()

# --- PRUEBA Y VALIDACIÓN ---
ciudades_prueba = ["  barcelona ", "MADRID", None]
resultados = [normalizar_texto(c) for c in ciudades_prueba]
print("Resultados de normalización:", resultados)
````

### 🟡 Ejercicio 4.3 — Función de Regla de Calidad de Datos *(Intermedio)*
**Contexto:** En un flujo ETL profesional, separamos las reglas de validación en funciones booleanas (`True` / `False`) para que cualquier compañera del equipo pueda leerlas y mantenerlas fácilmente.

**Tu tarea:**
1. Crea una función `es_registro_valido(fila)` que reciba un diccionario `fila` (con claves `"unidades"` y `"precio"`) y devuelva `True` **únicamente si**:
   * `"unidades"` es mayor que `0`, **Y**
   * `"precio"` no es `None`, **Y**
   * `"precio"` es mayor o igual que `0`.
2. Prueba tu función con los 3 diccionarios de prueba proporcionados abajo.

> 💡 **Pista:** Puedes devolver directamente el resultado de la expresión lógica: `return fila.get("unidades", 0) > 0 and ...`
> ✅ **Resultado esperado:** `True`, `False`, `False`

````python
def es_registro_valido(fila):
    # --- POR QUÉ ---
    # Usamos .get() en lugar de corchetes directos para que si alguna columna clave falta por error
    # en el diccionario origen, no rompa el código con un KeyError, sino que devuelva un valor por defecto.
    unidades = fila.get("unidades", 0)
    precio = fila.get("precio", None)

    # --- FILTROS DE CALIDAD ---
    # 1. Las unidades deben ser mayores a 0 (no admitimos registros con 0 o negativos de devoluciones en esta fase).
    # 2. El precio no puede ser un valor nulo (is not None).
    # 3. El precio debe ser un valor matemático no negativo (precio >= 0.0).
    return unidades > 0 and precio is not None and precio >= 0.0

# --- PRUEBA Y VALIDACIÓN ---
fila_ok = {"categoria": "Tech", "unidades": 2, "precio": 50.0}
fila_negativa = {"categoria": "Tech", "unidades": -1, "precio": 50.0}
fila_nula = {"categoria": "Tech", "unidades": 3, "precio": None}

print("¿Fila OK es válida?:", es_registro_valido(fila_ok))       # Esperado: True
print("¿Fila negativa es válida?:", es_registro_valido(fila_negativa)) # Esperado: False
print("¿Fila nula es válida?:", es_registro_valido(fila_nula))     # Esperado: False
````

### 🟠 Ejercicio 4.4 — Etapa `Transform` Modular *(Intermedio-Alto)*
**Contexto:** Ahora vas a reutilizar las funciones que creaste en los ejercicios **4.2** y **4.3** dentro de una nueva función llamada `transformar_datos(filas_crudas)`. ¡Esto es programación modular real!

**Tu tarea:**
1. Define `transformar_datos(filas_crudas)`.
2. Dentro de ella, crea una lista `datos_limpios = []` y recorre `filas_crudas`:
   * Usa `es_registro_valido(fila)` para filtrar las filas corruptas.
   * Si es válida, añade a `datos_limpios` un nuevo diccionario con:
     * `"categoria"`: usando `normalizar_texto(fila.get("categoria"))`
     * `"ingreso"`: `round(fila["unidades"] * fila["precio"], 2)`
3. Devuelve `datos_limpios` con `return`.
4. Ejecuta la función sobre `dataset_ventas_raw` e imprime el resultado.

> ✅ **Resultado esperado:** `[{'categoria': 'Tech', 'ingreso': 200.0}, {'categoria': 'Tech', 'ingreso': 150.0}]`

````python
def transformar_datos(filas_crudas):
    # --- POR QUÉ ---
    # Creamos un contenedor limpio para no mutar ni modificar el dataset original en memoria (buenas prácticas).
    datos_limpios = []

    for fila in filas_crudas:
        # --- POR QUÉ ---
        # Modularidad: Reutilizamos la función de auditoría del Ejercicio 4.3 para decidir si procesamos la fila.
        if es_registro_valido(fila):
            # --- POR QUÉ ---
            # Modularidad: Reutilizamos la función del Ejercicio 4.2 para limpiar el texto de la categoría sin repetir lógica.
            cat_limpia = normalizar_texto(fila.get("categoria"))

            # Calculamos el ingreso derivado de la transacción con redondeo estandarizado a 2 decimales
            ingreso_calculado = round(fila["unidades"] * fila["precio"], 2)

            # Generamos la nueva estructura estandarizada
            fila_limpia = {
                "categoria": cat_limpia,
                "ingreso": ingreso_calculado
            }
            datos_limpios.append(fila_limpia)

    return datos_limpios

# --- PRUEBA Y VALIDACIÓN ---
dataset_ventas_raw = [
    {"categoria": " tech ", "unidades": 2, "precio": 100.0},
    {"categoria": "hogar", "unidades": -1, "precio": 50.0},
    {"categoria": "TECH", "unidades": 1, "precio": 150.0},
    {"categoria": "moda", "unidades": 3, "precio": None},
]

ventas_procesadas = transformar_datos(dataset_ventas_raw)
print("Registros limpios procesados:", ventas_procesadas)
````

````python
def generar_reporte_etl(filas_crudas):
    # --- PASO 1: Extracción y Transformación (Modular) ---
    # Reutilizamos la función del Ejercicio 4.4 para obtener la lista de datos ya auditada y limpia.
    limpios = transformar_datos(filas_crudas)

    # --- PASO 2: Métricas de Control e Ingesta ---
    registros_entrada = len(filas_crudas)
    registros_validos = len(limpios)

    # --- POR QUÉ ---
    # Añadimos un bloque de control condicional para prevenir el error fatal de división por cero
    # (ZeroDivisionError) en caso de que el dataset de entrada llegue vacío.
    if registros_entrada > 0:
        tasa_calidad_pct = round((registros_validos / registros_entrada) * 100, 1)
    else:
        tasa_calidad_pct = 0.0

    # --- PASO 3: Consolidación Financiera ---
    facturacion_total = round(sum(item["ingreso"] for item in limpios), 2)

    # --- PASO 4: Agrupación analítica (GROUP BY) por categoría limpia ---
    facturacion_por_categoria = {}
    for item in limpios:
        cat = item["categoria"]
        # Usamos .get(cat, 0.0) para acumular el ingreso sumando desde cero si la categoría es nueva
        facturacion_por_categoria[cat] = round(facturacion_por_categoria.get(cat, 0.0) + item["ingreso"], 2)

    # --- PASO 5: Retorno de resultados ---
    # Devolvemos un reporte consolidado estructurado como diccionario para su fácil consumo por otras APIs.
    return {
        "registros_entrada": registros_entrada,
        "registros_validos": registros_validos,
        "tasa_calidad_pct": tasa_calidad_pct,
        "facturacion_total": facturacion_total,
        "facturacion_por_categoria": facturacion_por_categoria
    }

# --- PRUEBA Y VALIDACIÓN DEL PIPELINE COMPLETO ---
reporte_ejecutivo = generar_reporte_etl(dataset_ventas_raw)
print("Reporte Ejecutivo ETL Consolidado:")
print(reporte_ejecutivo)
````
