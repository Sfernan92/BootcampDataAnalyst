# Calculadora de indicadores de Marketing

````python
#datos

impresiones = 12000
clics = 420
inversion = 210
ingresos_genrados = 680

#calcula

# CTR porcentaje de impresiones que se convierten en clics, clics/impresiones*100
ctr = round(clics / impresiones * 100, 2)
print(f"CTR es {ctr} %.")
# CPC coste medio de cada clic, inversion/clic
cpc = round(inversion / clics, 2)
print(f"CPC es {cpc}.")
# Beneficio, ingresos-inversion
beneficio = round(ingresos_genrados - inversion, 2)
print(f"Beneficio es {beneficio}.")
#ROI retorno de la inversion en porcentaje, beneficio/inversion*100
roi= round(beneficio / inversion * 100, 2)
print(f"ROI es {roi} %.")


# PRUEBA DE PONER DECIMALES CON .2f

print("\nOtra forma de ver como oner los decimales:")

ctr = (clics / impresiones * 100)
print(f"\nCTR es {ctr: .2f} %.")

# PRUEBA DE PONER DECIMALES CON .2f
roi_pruebadecimales= (beneficio / inversion * 100)
print(f"ROI PRUEBA es {roi_pruebadecimales:.2f} %.")


````

# 💡 Explicación de los Métodos de Cadenas de Texto:

* **`.strip()`**:
  * **¿Qué hace?** Quita los espacios vacíos del inicio y del final del texto.
  * **Ejemplo:** `"  hola  ".strip()` se convierte en `"hola"`.

* **`.title()`**:
  * **¿Qué hace?** Pone la primera letra de cada palabra en mayúscula y todas las demás en minúscula.
  * **Ejemplo:** `"teclado MECÁNICO".title()` se convierte en `"Teclado Mecánico"`.

* **`.replace("valor_viejo", "valor_nuevo")`**:
  * **¿Qué hace?** Busca un texto y lo sustituye por otro.
  * **Ejemplo:** `"89.90€".replace("€", "")` elimina el símbolo de euro y nos devuelve `"89.90"`.


# 🚀 Otros métodos de texto muy utilizados:

* **`.upper()`**:
  * **¿Qué hace?** Convierte todo el texto a mayúsculas. Muy útil para estandarizar códigos de campaña o cupones.
  * **Ejemplo:** `"descuento10".upper()` se convierte en `"DESCUENTO10"`.

* **`.lower()`**:
  * **¿Qué hace?** Convierte todo el texto a minúsculas. Ideal para limpiar correos electrónicos o búsquedas.
  * **Ejemplo:** `"Sandra@Email.Com".lower()` se convierte en `"sandra@email.com"`.

* **`.split(separador)`**:
  * **¿Qué hace?** Corta o divide un texto en varias partes usando el separador que le indiques, y te devuelve una **lista** con los trozos.
  * **Ejemplo:** `"pantalón,camisa,zapatos".split(",")` se convierte en `['pantalón', 'camisa', 'zapatos']`.

* **`.join(lista)`**:
  * **¿Qué hace?** Hace lo contrario que `.split()`. Une los elementos de una lista de textos en una sola cadena, usando el texto que desees como conector.
  * **Ejemplo:** `"-".join(["2026", "10", "27"])` se convierte en `"2026-10-27"`.

* **`.startswith(texto)` / `.endswith(texto)`**:
  * **¿Qué hace?** Comprueba si el texto empieza o termina con una palabra o letra concreta. Devuelve `True` o `False`.
  * **Ejemplo:** `"Campaña_Navidad".startswith("Campaña")` devuelve `True`.

### 📚 Estructuras de Datos: Listas, Tuplas y Diccionarios

En Python, cuando queremos guardar más de un dato en una sola variable, utilizamos colecciones. Las tres más importantes son:

---

#### 1. Listas `[]` (Mutables / Modificables)
Las listas son colecciones **ordenadas** de elementos que **se pueden modificar** (añadir, quitar o cambiar elementos) después de haber sido creadas. Se definen utilizando corchetes `[]`.

* **Ejemplo de uso:** Una lista de compras, nombres de alumnos, o tus últimas campañas de marketing.

```python
# Crear una lista
frutas = ["manzana", "plátano", "naranja"]

# Modificar un elemento
frutas[1] = "pera"  # Ahora el plátano es una pera

# Añadir un elemento nuevo
frutas.append("fresa")

print(frutas)  # Resultado: ['manzana', 'pera', 'naranja', 'fresa']
```

##### 🛠️ Métodos de listas más comunes:

* **`lista.append(elemento)`**: Añade un elemento al final de la lista.
  ```python
  nombres = ["Ana", "Luis"]
  nombres.append("Pedro") # ['Ana', 'Luis', 'Pedro']
  ```

* **`lista.insert(posición, elemento)`**: Inserta un elemento en la posición (índice) que tú decidas. Los elementos siguientes se desplazan.
  ```python
  nombres = ["Ana", "Luis"]
  nombres.insert(1, "Sofía") # ['Ana', 'Sofía', 'Luis'] (Sofía entra en la posición 1)
  ```

* **`lista.remove(elemento)`**: Busca y elimina el primer elemento que coincida con el valor que le pases.
  ```python
  nombres = ["Ana", "Luis", "Ana"]
  nombres.remove("Ana") # ['Luis', 'Ana'] (elimina solo el primer 'Ana')
  ```

* **`lista.pop(posición)`**: Elimina el elemento de la posición indicada y te lo "devuelve" (por si quieres guardarlo en otra variable). Si no le pones número, elimina y devuelve el **último**.
  ```python
  nombres = ["Ana", "Luis", "Pedro"]
  eliminado = nombres.pop(1) # Quita a 'Luis'
  # nombres queda: ['Ana', 'Pedro'] | eliminado vale: 'Luis'
  ```

* **`lista.sort()`**: Ordena la lista automáticamente (alfabéticamente para texto, o de menor a mayor para números).
  ```python
  numeros = [5, 1, 8]
  numeros.sort() # [1, 5, 8]
  ```

* **`len(lista)`**: No es un método de lista en sí, sino una función global muy usada para saber cuántos elementos tiene una lista.
  ```python
  nombres = ["Ana", "Luis"]
  print(len(nombres)) # Resultado: 2
  ```

---

#### 2. Tuplas `()` (Inmutables / No modificables)
Las tuplas son colecciones **ordenadas** de elementos, pero **no se pueden modificar** una vez creadas. Son más rápidas y seguras que las listas porque protegen tus datos de cambios accidentales. Pueden tener elementos de distintos tipos. Se definen utilizando paréntesis `()`.

* **Ejemplo de uso:** Coordenadas GPS (latitud, longitud), meses del año, o valores constantes que nunca deben cambiar.

```python
# Crear una tupla
coordenadas = (40.4167, -3.7037)

# Si intentas hacer: coordenadas[0] = 50.0, Python te dará un ERROR porque no se puede modificar.
print(coordenadas)
```

---

#### 3. Diccionarios `{}` (Clave-Valor)
Los diccionarios son colecciones **sin un orden específico** que guardan información en parejas de **Clave: Valor**. En lugar de acceder a los elementos por su posición (como `0, 1, 2`), accedes a ellos mediante una palabra clave (llave). Se definen utilizando llaves `{}`.

* **Ejemplo de uso:** Información de un usuario, características de un producto o métricas de una campaña.

```python
# Crear un diccionario
campaña_detalles = {
    "nombre": "Verano",
    "clics": 420,
    "activo": True
}

# Acceder a un valor usando su clave
print(campaña_detalles["nombre"])  # Resultado: Verano

# Modificar un valor o añadir uno nuevo
campaña_detalles["clics"] = 450  # Modificamos los clics
campaña_detalles["presupuesto"] = 500  # Añadimos un dato nuevo

print(campaña_detalles)
```

##### 🛠️ Métodos de diccionarios más comunes:

* **`diccionario.keys()`**: Te devuelve una lista con todas las **claves** (los nombres o etiquetas) del diccionario.
  ```python
  datos = {"nombre": "Ana", "edad": 25}
  print(datos.keys()) # Resultado: dict_keys(['nombre', 'edad'])
  ```

* **`diccionario.values()`**: Te devuelve una lista con todos los **valores** (los datos guardados) del diccionario, sin las claves.
  ```python
  datos = {"nombre": "Ana", "edad": 25}
  print(datos.values()) # Resultado: dict_values(['Ana', 25])
  ```

* **`diccionario.items()`**: Devuelve una lista con parejas de elementos (clave, valor) en formato de tuplas. Es ideal cuando quieres recorrer un diccionario entero con un bucle.
  ```python
  datos = {"nombre": "Ana", "edad": 25}
  print(datos.items()) # Resultado: dict_items([('nombre', 'Ana'), ('edad', 25)])
  ```

* **`diccionario.get(clave, valor_por_defecto)`**: Sirve para obtener el valor de una clave **de forma segura**. Si la clave no existe, en lugar de dar un error que detenga tu programa, te devolverá `None` o el valor que tú decidas por defecto.
  ```python
  datos = {"nombre": "Ana"}
  # Intentar acceder a una clave que no existe:
  # print(datos["correo"]) -> Daría un ERROR catastrófico.
  print(datos.get("correo")) # Resultado: None (no da error)
  print(datos.get("correo", "No especificado")) # Resultado: "No especificado"
  ```

* **`diccionario.update(otro_diccionario)`**: Permite actualizar el diccionario añadiendo o modificando varias parejas de clave-valor a la vez.
  ```python
  datos = {"nombre": "Ana", "edad": 25}
  datos.update({"edad": 26, "ciudad": "Madrid"})
  # Ahora datos es: {"nombre": "Ana", "edad": 26, "ciudad": "Madrid"}
  ```

---

### ⚖️ Resumen rápido:

| Estructura | Símbolo | ¿Se puede cambiar? (Mutable) | ¿Tiene orden? | ¿Cómo se accede? |
| :--- | :---: | :---: | :---: | :--- |
| **Lista** | `[]` | Sí | Sí | Por posición (`lista[0]`) |
| **Tupla** | `()` | No | Sí | Por posición (`tupla[0]`) |
| **Diccionario** | `{}` | Sí | No | Por clave (`diccionario["clave"]`) |

### 🛠️ Diferencia entre Función y Método en Python

Aunque ambos son bloques de código diseñados para realizar una tarea específica, la diferencia principal radica en **a quién pertenecen** y **cómo se ejecutan**.

---

#### 1. ¿Qué es una Función?
Una **función** es un bloque de código independiente. Existe por sí sola en tu programa y no depende de ningún objeto específico para ser utilizada.

* **¿Cómo se llama?** Directamente por su nombre, pasando los datos (argumentos) dentro de los paréntesis: `nombre_funcion(dato)`.
* **Analogía:** Imagina una herramienta universal e independiente como un *destornillador* o unas *tijeras*. Puedes usarlos con cualquier cosa directamente.

**Ejemplos de funciones integradas (built-in):**
```python
# len() es una función: le pasamos la lista como parámetro
frutas = ["manzana", "pera"]
largo = len(frutas)

# print() es una función independiente
print("Hola")
```

---

#### 2. ¿Qué es un Método?
Un **método** es esencialmente una función, pero **está ligada y pertenece a un tipo de objeto** (como a un String, una Lista, un Diccionario, etc.). Un método está diseñado para interactuar con los datos del objeto que lo posee.

* **¿Cómo se llama?** Se utiliza la "nomenclatura del punto": `objeto.nombre_metodo()`.
* **Analogía:** Imagina las funciones automáticas de un *teléfono móvil* (como `llamar()` o `enviar_mensaje()`). No puedes usarlas solas en el aire, necesitas obligatoriamente el objeto *teléfono* para ejecutarlas (`telefono.llamar()`).

**Ejemplos de métodos:**
```python
texto = "hola"
# .upper() es un método que pertenece exclusivamente a los objetos de texto (strings)
texto_mayuscula = texto.upper()

frutas = ["manzana"]
# .append() es un método que pertenece exclusivamente a las listas
frutas.append("pera")
```

---

### ⚖️ Comparativa Directa:

| Característica | Función | Método |
| :--- | :--- | :--- |
| **Independencia** | Es autónoma, vive sola. | Pertenece a una clase/objeto específico. |
| **Sintaxis de uso** | `funcion(objeto)` | `objeto.metodo()` |
| **Relación** | Opera **sobre** los datos que le pasas. | Es una capacidad que **tiene** el propio objeto. |
| **Ejemplo** | `len("hola")` | `"hola".upper()` |

# 🔄 Introducción a los Bucles (Loops) en Python

En programación, un **bucle** es una estructura que nos permite **repetir un bloque de código** varias veces sin tener que escribirlo de nuevo. En el análisis de datos son fundamentales, ya que solemos aplicar la misma operación a miles de filas de datos, clientes o productos.

En Python existen dos tipos principales de bucles: **`for`** y **`while`**.

---

#### 1. El Bucle `for` (Para cada elemento...)
Se utiliza cuando **sabemos de antemano cuántas veces** queremos repetir la acción. Es ideal para recorrer colecciones de datos (como listas, tuplas o diccionarios).

* **Analogía:** Imagina que tienes una caja con 5 manzanas y las vas sacando una a una para lavarlas. El proceso termina cuando la caja se queda vacía.

##### Sintaxis básica:
```python
frutas = [’manzana’, ’pera’, ’naranja’]

# "Para cada fruta en la lista de frutas..."
for fruta in frutas:
    print(f"Lavando: {fruta}")
```
*La variable `fruta` es una variable temporal que toma el valor de cada elemento de la lista en cada vuelta del bucle.*

##### Uso muy común: la función `range()`
Si solo queremos repetir algo un número fijo de veces, usamos `range(inicio, fin)`:
```python
# Repite el código 3 veces (del 0 al 2)
for i in range(3):
    print(f"Vuelta número {i}")
```

---

#### 2. El Bucle `while` (Mientras se cumpla la condición...)
Se utiliza cuando **no sabemos exactamente cuántas veces** se va a repetir el bucle, sino que este continuará ejecutándose **mientras una condición sea verdadera (`True`)**.

* **Analogía:** Imagina que estás inflando un globo. Tu condición para seguir inflando es: *"mientras el globo no esté lleno"*. En cuanto se llena, te detienes.

##### Sintaxis básica:
```python
energia = 3

# "Mientras la energía sea mayor que 0..."
while energia > 0:
    print(f"Corriendo... Energía restante: {energia}")
    energia = energia - 1  # ¡Es súper importante reducirla para que el bucle no sea infinito!

print("¡Me he quedado sin energía!")
```

⚠️ **¡Cuidado con los bucles infinitos!** Si la condición del `while` nunca se vuelve falsa, el ordenador se quedará ejecutando el bucle para siempre y el programa se congelará.

| Tipo de Bucle | ¿Cuándo usarlo? | Ejemplo práctico |

# 🚦 Sentencias Condicionales en Python: `if`, `elif` y `else`

Las estructuras condicionales nos permiten **tomar decisiones** en nuestro código. Le dicen al ordenador: *"Si pasa esto, haz una cosa; si pasa esto otro, haz algo diferente"*.

Esto es clave en análisis de datos, por ejemplo, para clasificar clientes según sus compras, comprobar si una campaña está activa o etiquetar si un precio es caro o barato.

---

#### 🧩 La Estructura Básica

* **`if` (Si...)**: Evalúa una condición. Si es verdadera (`True`), ejecuta el código de su interior.
* **`elif` (Sino si...)**: Es la abreviatura de *else if*. Se usa cuando la primera condición no se cumplió, pero queremos comprobar otra nueva.
* **`else` (Sino...)**: Es el "cajón de sastre". Se ejecuta automáticamente si ninguna de las condiciones anteriores fue verdadera.

##### 💡 Analogía de la vida real:
Imagina que vas a salir de casa y miras al cielo:
* **Si (`if`)** está lloviendo 🌧️ ➡️ Cojo un paraguas.
* **Sino si (`elif`)** está nublado ☁️ ➡️ Cojo una chaqueta.
* **Sino (`else`)** (hace sol) ☀️ ➡️ Cojo gafas de sol.

---

#### 💻 Ejemplo en Python:

```python
presupuesto = 150

if presupuesto > 500:
    print("Tenemos un presupuesto Alto para la campaña.")
elif presupuesto >= 100:
    print("Tenemos un presupuesto Medio.")
else:
    print("El presupuesto es Bajo. ¡Hay que optimizar!")
```

#### ⚠️ Regla de Oro: La Indentación (El espacio)
En Python, no usamos llaves `{}` para agrupar el código de una condición. En su lugar, usamos **dos puntos `:`** al final de la condición y dejamos **4 espacios de sangría (indentación)** en la siguiente línea.
Todo lo que esté alineado con esos 4 espacios pertenecerá a esa condición.

* `>` (Mayor que) y `<` (Menor que)
* `>=` (Mayor o igual) y `<=` (Menor o igual)
* `==` (Igual a) ⚠️ *¡Atención! Se usan dos signos de igual para comparar, un solo `=` sirve para asignar variables.*
* `!=` (Diferente de)
