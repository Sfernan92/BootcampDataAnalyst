Este notebook usa el CSV sintético. Ejecuta las celdas en orden y después interpreta.


```python
from pathlib import Path
import csv
from decimal import Decimal

base = Path.cwd()
if not (base / "datos/operaciones.csv").exists():
    base = base.parent
ruta = base / "datos" / "operaciones.csv"
with ruta.open(encoding="utf-8", newline="") as f:
    filas = list(csv.DictReader(f))
print("Filas del CSV:", len(filas))
```

```python
total = len(filas)
completadas = [f for f in filas if f["estado"] == "completada"]
canceladas = [f for f in filas if f["estado"] == "cancelada"]
devueltas = [f for f in filas if f["estado"] == "devuelta"]
ingresos = sum((Decimal(f["importe_eur"]) for f in completadas), Decimal("0.00"))
print(f"Total: {total} | Completadas: {len(completadas)} | Canceladas: {len(canceladas)} | Devueltas: {len(devueltas)}")
print(f"Ingresos completados (EUR): {ingresos:.2f}")
```
